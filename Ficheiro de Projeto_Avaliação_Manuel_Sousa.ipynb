```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

texto= 'Na disciplina de multimédia, vamos aprender algumas coisas interessantes'

as_minhas_stopwords= set(['Na' , 'de' , 'vamos' , 'algumas'])

#--------------------Configurar e gerar a nuvem de palavras apartir de palavras-------------------------

nuvem= WordCloud(
    width=1000, 
    height=550, 
    background_color='white', 
    colormap='plasma',
    stopwords=as_minhas_stopwords).generate(texto)

#--------------------Exibir a nuvem de palavras atraves de módulo Matplotlib-------------------------

plt.figure(figsize=(12, 7))
plt.imshow(nuvem, interpolation='bilinear')
plt.axis('off')
plt.show()

print(nuvem)
```
